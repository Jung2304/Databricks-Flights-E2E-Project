### CREATING PARAMETERS FOR DYNAMIC SOLUTION

```sql
%python

# SOURCE SCHEMA
dbutils.widgets.text("source_schema", "silver")

# SOURCE TABLE
dbutils.widgets.text("source_table", "")

# TARGET SCHEMA
dbutils.widgets.text("target_schema", "gold")

# TARGET TABLE
dbutils.widgets.text("target_table", "")

# KEY COLUMNS
dbutils.widgets.text("key_cols", '["placeholder"]')

# CDC COLUMN (Load data incrementally)
dbutils.widgets.text("cdc_col", "modifiedDate")

# BACK-DATED REFRESH
dbutils.widgets.text("backdated_refresh", "")

# SURROGATE KEY
dbutils.widgets.text("surrogate_key", "")
```

```sql
%python

# Read the current widget values
parameter_names = [
    "source_schema",
    "source_table",
    "target_schema",
    "target_table",
    "key_cols",
    "cdc_col",
    "backdated_refresh",
    "surrogate_key"
]

# Publish each value for downstream tasks
for name in parameter_names:
    dbutils.jobs.taskValues.set(
        key=name,
        value=dbutils.widgets.get(name)
    )
```
